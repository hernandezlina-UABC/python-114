```python
# Functions
def my_function():
    a = 10
    b = 3
    print (a + b)

def my_function_with_parameters(a, b):
    print (a + b)
    
def my_function_with_return():
    return 10 + 3
    
def my_complete_function(a, b):
    return a + b

my_function()
my_function_with_parameters(10,3)
result = my_function_with_return()
print(result)
print(my_complete_function(10,3))

def function_with_args(*args):
    print("Type:", type(args))
    print("First Argument:", args[0])
    print("Second Argument:", args[1])
    print("Third Argument:", args[2])
    
function_with_args("One", "Two", "Three")

def function_with_kwargs(**kwargs):
    print("Type:", type(kwargs))
    print("My name is " + kwargs["name"])
    print("My age is " + str(kwargs["age"]))
    
function_with_kwargs(name="Lina", age=26, this_is_a_real_variable_i_swear="Hello")

def combined_function(name, *args, **kwargs):
    # Order Matters:
    # 1. Positional parameter
    # 2. *args
    # 3. **kwargs
    print(name)
    print(args)
    print(kwargs)
    
combined_function("Luis", "Lina", "JL", age=26, is_activate=False)
```

# OOP (Object Oriented Programing)
Python is an object-oriented programming, allowing us to structure code using classes and objects for a better organization and reusability.

## Advantages
- Provides clear structure to programs
- Makes code easier to mantain, reuse and debug
- Helps keep your code DRY (Do not Repeat Yourself principle)
- Allows to rebuild reusable applications with less code

## Classes and Objects
A class defines what an object should look like and it's behavior  
An object is created based on that class  
For example:  
    Class -> Fruit  
    Objects -> Apple, Banana, Cherry  
    Class -> Car  
    Objects -> Audi, Tesla, Ford  

#### The four pillars of OOP
1. Encapsulation
2. Abstraction
3. Inheritance
4. Polymorphism

Example of a class:
    class PostListView(ListView)


```python
class Person:
    # Constructor
    def __init__(self, name, last_name, age):
        self.name = name
        self.last_name = last_name
        self.age = age
        self.is_active = False

    def greeting(self):
        return f"Hello from {self.name}"

    def __str__(self):
        return f"Person (name = {self.name}, last_name = {self.last_name}, age = {self.age}, is_active = {self.is_active})"

p1 = Person("Lina", "Hernandez", 26)
p1.is_active = True
print(p1)
print(p1.greeting())

p2 = Person("Tom", "Holland", 29)
print(p2)
print(p2.greeting())
```
