## Stacks
A stack is an ordered collection of elements where items are added and removed fron one end, commonly refered to as the "top". The ordering principle is referred to as LIFO (Last In, First Out)

```python
# Simplified Version (relying on built-ins functions)

class Stack:
    def __init__(self):
        self.items = []

    def push(self, elements):
        self.items.append(elements)
    
    def push(self, *args):
        for item in args:
            self.items.append(item)        
        
    def pop(self): # getting out elements (first the last value)
        return self.items.pop()
    
    def peek(self):
        return self.items[-1]
    
    def is_empty(self):
        return self.items == []
    
    def size(self):
        return len(self.items)
    
    def __str__(self):
        return f"Stack - {self.items}"
```

```python
s1 = Stack()
print(s1)
print(s1.is_empty())

s1.push("A")
s1.push("B")
s1.push("C")
print(s1.size())
print(s1.peek())
print(s1.is_empty())
top = s1.pop()
print(s1.size())
print(top)
s1.push("C", "D", "E")
print(s1)
```

```python
def invert_str(str):
    s2 = Stack()
    result = ""
    
    for char in str:
        s2.push(char)
        
    while not s2.is_empty():
        result += s2.pop()
        
    return result
invert_str("Rafael")
invert_str()
```
