#### Mini Challenge 1: Largest Num
Create a function that will help us to determine the largest number from a list of numbers

Criteria: 
1. The function should receive a list and return the largest number
2. You can test it with a list of any values from your selection

```python
def identify_largest_number(numbers):
    largest_num = 0
    for i in numbers:
        if i > largest_num:
            largest_num = i
    return largest_num

numbers = [38, 1, 40, 10, 6, 20, 100, 230, -1]
print(f"The largest number is: {identify_largest_number(numbers)}")
```

```python
def find_largest_number(numbers):
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number
    
    return largest

print(f"The Largest number is: {identify_largest_number(numbers)}")
```

## Mini Challenge 2: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagram of each other. False otherwise.

    Example:
    1. heart and earth
    2. scar and cars
    3. star and rats

*Criteria:*
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be exactly the same length (in chars)
    3. AI and sorted function is not allowed

```python
def are_anagrams(a, b):
    counter = 0
    if len(a) == len(b):        
        for letter1 in a:
            for letter2 in b:
                if letter1 == letter2:
                    counter += 1
        if len(a) != counter:
            return False
        else: 
            return True
    else:
        return False

are_anagrams("art", "rat")
```

#### Assignment 2
Create a class called calculator that requires two attributes and contains four functions: add, subtract, multiply and divide.
Also add the str function to print the object.

Example:

c1 = Calculator(5,2)

c1.add() -> 7

c1.multiply() -> 10

c1.subtract() -> 3


```python
class Calculator:
    def __init__(self, first_number, second_number):
        self.first_number = first_number
        self.second_number = second_number
    
    def add(self):
        sum_result = self.first_number + self.second_number
        return sum_result
    
    def multiply(self):
        multiply_result = self.first_number * self.second_number
        return multiply_result
    
    def subtract(self):
        subtract_result = self.first_number - self.second_number
        return subtract_result
    
    def divide(self):
        divide_result = self.first_number / self.second_number
        return divide_result
    
    def __str__(self):
        return f"You are doing a calculation with {self.first_number} and {self.second_number}.. sum = {self.add()}, multiply = {self.multiply()}, subtraction = {self.subtract()}, divide = {self.divide()})"
   
c1 = Calculator(10, 2)
print(c1)
print(c1.add())
print(c1.multiply())
print(c1.subtract())
print(c1.divide())
```
