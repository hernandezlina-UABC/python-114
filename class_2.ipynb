#### Mini Challenge 1: Largest Num
Create a function that will help us to determine the largest number from a list of numbers

Criteria: 
1. The function should receive a list and return the largest number
2. You can test it with a list of any values from your selection

```python
def identify_largest_number(numbers):
    largest_num = 0
    for i in numbers:
        if i > largest_num:
            largest_num = i
    return largest_num

numbers = [38, 1, 40, 10, 6, 20, 100, 230, -1]
print(f"The largest number is: {identify_largest_number(numbers)}")
```

```python
def find_largest_number(numbers):
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number
    
    return largest

print(f"The Largest number is: {identify_largest_number(numbers)}")
```

## Mini Challenge 2: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagram of each other. False otherwise.

    Example:
    1. heart and earth
    2. scar and cars
    3. star and rats

*Criteria:*
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be exactly the same length (in chars)
    3. AI and sorted function is not allowed

```python
def are_anagrams(a, b): # O(1)
    counter = 0
    if len(a) == len(b): # O(1)      
        for letter1 in a: # O(n)
            for letter2 in b: # O(n) -> O(n2)
                if letter1 == letter2: # O(1)
                    counter += 1 # O(1)
        if len(a) != counter: # O(1)
            return False
        else: 
            return True
    else:
        return False

are_anagrams("art", "rat")
```

```python
def are_anagrams_v2(w1, w2):
    w1 = w1.lower().replace(" ", "")
    w2 = w2.lower().replace(" ", "")
    dict1 = {}
    dict2 = {}
    if len(w1) == len(w2):
        for letter in w1: # O(n)
            if dict1.get(letter) == None:
                dict1[letter] = 1
            else:
                dict1[letter] += 1
                
        for letter in w2: # O(n)
            if dict2.get(letter) == None:
                dict2[letter] = 1
            else:
                dict2[letter] += 1
                
        print(dict1)
        print(dict2)
        return dict1 == dict2
    return False

are_anagrams_v2("cars", "scar")
```

#### Assignment 2
Create a class called calculator that requires two attributes and contains four functions: add, subtract, multiply and divide.
Also add the str function to print the object.

Example:

c1 = Calculator(5,2)

c1.add() -> 7

c1.multiply() -> 10

c1.subtract() -> 3


```python
class Calculator:
    def __init__(self, first_number, second_number):
        self.first_number = first_number
        self.second_number = second_number
    
    def add(self):
        sum_result = self.first_number + self.second_number
        return sum_result
    
    def multiply(self):
        multiply_result = self.first_number * self.second_number
        return multiply_result
    
    def subtract(self):
        return self.first_number - self.second_number
    
    def divide(self):
        if self.second_number == 0
            raise ArithmeticError("Element can't be devide by 0")
        return self.first_number / self.second_number
    
    def __str__(self):
        return f"You are doing a calculation with {self.first_number} and {self.second_number}.."
   
c1 = Calculator(10, 2)
print(c1)
print(c1.add())
print(c1.multiply())
print(c1.subtract())
print(c1.divide())
```
